# Linearise the freezing temperature

In this notebook we adapt the linear freezing temperature ($Tf = \lambda_1*S + \lambda_2 + \lambda_3*p$) by fitting the coefficients ($\lambda_1$, $\lambda_2$, $\lambda_3$) over the T-p ranges of our study (0 to 1 g/kg and 0 to 750 dbar).

We also compute the value of constant water specific heat capacity at the surface for our reference temperature and salinity (4°C and 0.5 g/kg, respectively).

```python
#import the libraries needed
import gsw
import numpy as np
from scipy.optimize import curve_fit
```

### Freezing temperature

**function to fit**

```python
def lin_Tf(xdata,lambda1, lambda2,lambda3):
    """
    Calculate the freezing temperature using equation: Tf = lambda1*S + lambda2 + lambda3*p.

    Parameters
    ----------
    xdata : vectors containing the salinity and pressure values
        xdata[0,:] = Absolute Salinity (g/kg)
        xdata[1,:] = pressure (dbar)
    lambda1, lambda2, lambda3 : coefficients

    Returns
    -------
    Tf : freezing temperature (°C)
    """  
    Tf=lambda1*xdata[0,:]+lambda2+lambda3*xdata[1,:]
    return Tf
```

**define xdata and ydata for the fit**

```python
#fit ranges
SA_min = 0 ; SA_max = 1 #g/kg
p_min = 0 ; p_max = 750 #dbar
```

```python
#number of values of the fit
taille=550
```

```python
#create xdata
xdata=np.ones((2,taille**2))*np.nan
xdata[0,:]=np.tile(np.linspace(SA_min,SA_max,taille),taille)
xdata[1,:]=np.repeat(np.linspace(p_min,p_max,taille),taille)
```

```python
#create ydata
ydata=gsw.CT_freezing(xdata[0,:],xdata[1,:],0)
```

**fitting**

```python
Tf_fit=curve_fit(lin_Tf,xdata,ydata,full_output=True)
```

```python
print(Tf_fit[0])
```

**fit evaluation**

```python
Y_Tf_fit=lin_Tf(xdata,-0.0572,0.0197,-7.5519e-4)
```

```python
RMSE_Tf_fit=np.sqrt(np.mean((Y_Tf_fit-ydata)**2))
```

```python
print("RMSE :", RMSE_Tf_fit)
```
